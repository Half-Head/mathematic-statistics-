```python
import numpy as np
from scipy.stats import t as Student
```

```python
group1 = np.array([83, 85])
group2 = np.array([84, 85, 85, 86, 86, 87])
group3 = np.array([86, 87, 87, 87, 88, 88, 88, 88, 88, 89, 90])
group4 = np.array([89, 90, 90, 91])
group5 = np.array([90, 92])
```

# a

```python
gr1 = np.array([[1, 0, 0, 0, 0]] * 2)
gr2 = np.array([[0, 1, 0, 0, 0]] * len(group2))
gr3 = np.array([[0, 0, 1, 0, 0]] * len(group3))
gr4 = np.array([[0, 0, 0, 1, 0]] * len(group4))
gr5 = np.array([[0, 0, 0, 0, 1]] * len(group5))

y = np.concat([group1, group2, group3, group4, group5])

psi = np.vstack([gr1, gr2, gr3, gr4, gr5])

F = psi.T @ psi
b = np.linalg.inv(F) @ psi.T @ y

n = psi.shape[0]
p = psi.shape[1]

e = y - psi @ b
RSS = e.T @ e

print (f'Уравнение линейной регресии: y = {b[0]} * x1 + {b[1]} * x2 + {np.round(b[2], 2)} * x3 + {b[3]} * x4 + {b[4]} * x5 + e')
```

```python
for i in range (5):
    delta_value = (b[i] * np.sqrt(n - p)) / (RSS * np.linalg.inv(F)[i, i])
    p_value = 2 * (1 - Student.cdf(abs(delta_value), n - p))

    print (f'p_value = {p_value} ', end='')
    if p_value < 0.05:
        print (f'b{i+1} значимый')
    else:
        print (f'b{i+1} незначимый')
```

# b

```python
pairs = []

for i in range (5):
    for j in range (i+1, 5):
        if i == j:
            continue
        
        delta_value = (b[i] - b[j]) * np.sqrt(n - p) / np.sqrt(RSS * (np.linalg.inv(F)[i, i] + np.linalg.inv(F)[j, j] - 2 * np.linalg.inv(F)[i, j]))

        p_value = (2 * (1 - Student.cdf(abs(delta_value), n - p)))
        pairs.append((i, j, p_value))

pairs = sorted(pairs, key=lambda x: x[2])


for i in range (len(pairs)):
    if pairs[i][2] < (0.05 / (len(pairs) - i)):
        print(f'b{pairs[i][0]} и b{pairs[i][1]} статистически различны') 
    else:
        print(f'b{pairs[i][0]} и b{pairs[i][1]} статистически одинаковые') 
```
